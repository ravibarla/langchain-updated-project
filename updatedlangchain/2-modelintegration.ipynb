model integration with openAI,google gemini and groq

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

```

integrate google gemini integration

```python

from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")

model=init_chat_model(
    "google_genai:gemini-3.6-flash"
)

model.invoke("why do the parrot talks ?")
```


chat google gen ai

```python
import os 
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model= "gemini-3.6-flash")
response=model.invoke("why do parrot talks")
response
```

integrate groq integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3.6-27b")
response=model.invoke("why do parrots talk?")
response
```

Streaming

```python
for chunk in model.stream("why do parrot have colorful feathers ?"):
    print(chunk.text,end="|",flush=True)
```

Batch

```python
responses=model.batch([
"why do parrot have colorful feathers ?",
"How do Airoplane flies ?",
"what is quantum computing ?"
],config={"max_concurrency":5})

for response in responses:
    print(response)
```
